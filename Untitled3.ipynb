```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

pd.set_option('display.max_colwidth', None)
```

```python
mk = pd.DataFrame([
    {"kode": "MK01", "nama": "Pengembangan Aplikasi Web", "tag": "web backend frontend pemrograman"},
    {"kode": "MK02", "nama": "Data Mining", "tag": "data analitik statistik machine-learning"},
    {"kode": "MK03", "nama": "Keamanan Jaringan", "tag": "jaringan keamanan sistem infrastruktur"},
    {"kode": "MK04", "nama": "Machine Learning Lanjut", "tag": "machine-learning data analitik ai"},
    {"kode": "MK05", "nama": "Pengembangan Aplikasi Mobile", "tag": "mobile frontend pemrograman ui-ux"},
    {"kode": "MK06", "nama": "Manajemen Basis Data Lanjut", "tag": "data basis-data sistem backend"},
    {"kode": "MK07", "nama": "Cloud Computing", "tag": "infrastruktur jaringan sistem cloud"},
    {"kode": "MK08", "nama": "Interaksi Manusia dan Komputer", "tag": "ui-ux frontend desain pengguna"},
])
mk
```

```python
mahasiswa = {
    "nama": "ayu",
    "mk_pernah_diambil": ["MK02"],  # sudah pernah ambil, tidak direkomendasikan ulang
    "profil_minat": "data analitik machine-learning statistik ai"
}
mahasiswa
```

```python
corpus = list(mk["tag"]) + [mahasiswa["profil_minat"]]

vectorizer = CountVectorizer()
matrix = vectorizer.fit_transform(corpus)

mk_vectors = matrix[:-1]        # vektor untuk semua MK
mhs_vector = matrix[-1]         # vektor profil mahasiswa (baris terakhir)

skor = cosine_similarity(mhs_vector, mk_vectors)[0]
mk["skor_kecocokan"] = (skor * 100).round(1)
mk.sort_values("skor_kecocokan", ascending=False)
```

```python
def rekomendasikan_mk(mahasiswa, mk_df, top_n=3):
    hasil = mk_df[~mk_df["kode"].isin(mahasiswa["mk_pernah_diambil"])].copy()
    hasil = hasil.sort_values("skor_kecocokan", ascending=False).head(top_n)
    return hasil[["kode", "nama", "skor_kecocokan"]].reset_index(drop=True)

top_rekomendasi = rekomendasikan_mk(mahasiswa, mk, top_n=3)
print(f"Rekomendasi MK pilihan untuk {mahasiswa['nama']}:")
top_rekomendasi
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.barh(top_rekomendasi["nama"], top_rekomendasi["skor_kecocokan"], color="#2F5496")
plt.xlabel("Skor Kecocokan (%)")
plt.title(f"Top-{len(top_rekomendasi)} Rekomendasi MK Pilihan\nuntuk {mahasiswa['nama']}")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig("grafik_rekomendasi.png", dpi=120)
plt.show()
```
